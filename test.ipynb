```python
import unpack
from unpack.viz import CircuitGraph
from IPython.display import SVG, display

tracer = unpack.Tracer("gpt2", device="cuda:0")

text = "When Mary and John went to the store, John gave a drink to"

# Trace from target
result = tracer.trace(text, target=" Mary", distractor=" John", config="kqv_aligned")

# Top 5 paths from target
graph = CircuitGraph.from_tracer(tracer)
graph.tokens = result.tokens
graph.add_paths(result.paths[:5])
display(SVG(graph.render_svg()))

# Reroot at a name mover — see what feeds it
prep, cfg = tracer.prepare(text, target=" Mary", distractor=" John", config="kqv_aligned")
nm_result = tracer.trace_from_prep(prep, cfg, root="attn_9_head_9")

graph2 = CircuitGraph.from_tracer(tracer)
graph2.tokens = result.tokens
graph2.add_paths(nm_result.paths[:5])
display(SVG(graph2.render_svg()))
```
